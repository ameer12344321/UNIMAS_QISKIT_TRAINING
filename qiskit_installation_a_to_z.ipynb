# Qiskit Installation Guide (A to Z)

This notebook gives a complete beginner-friendly path to install Qiskit from scratch using Anaconda and Jupyter Notebook, then verify everything works correctly.

## A. Download and Install Anaconda

1. Go to the official Anaconda website: https://www.anaconda.com/download
2. Download the installer for your operating system (Windows/macOS/Linux).
3. Run the installer with default options (recommended for beginners).
4. After installation, open **Anaconda Navigator** or **Anaconda Prompt**.

## B. Create a Clean Python Environment (Recommended)

Use a dedicated environment for Qiskit to avoid package conflicts.

Run these commands in **Anaconda Prompt** (Windows) or terminal (macOS/Linux):

```bash
conda create -n qiskit-env python=3.11 -y
conda activate qiskit-env
```

## C. Install Jupyter Notebook

If Jupyter is not already available in your environment, install it:

```bash
conda install -y jupyter
```

Then launch Jupyter Notebook:

```bash
jupyter notebook
```

## D. Install Qiskit

Inside the same activated environment, install Qiskit with pip:

```bash
pip install qiskit
```

> Tip: You can also install in notebook cells using `%pip install qiskit`, but terminal installation is usually cleaner.

## E. Check Python and Qiskit Versions

Run the following cell to confirm your Python executable and Qiskit version:

````python
import sys
import qiskit

print('Python executable:', sys.executable)
print('Python version   :', sys.version.split()[0])
print('Qiskit version   :', qiskit.__version__)
````

## F. Run a Basic Qiskit Import Test

This confirms core Qiskit modules are importable:

````python
from qiskit import QuantumCircuit

qc = QuantumCircuit(2, 2)
qc.h(0)
qc.cx(0, 1)
qc.measure([0, 1], [0, 1])

print('QuantumCircuit created successfully!')
print(qc)
````

## G. Simulate a Circuit (Functional Test)

Run a simulator test to ensure Qiskit works end-to-end. If needed, install the simulator package:

```bash
pip install qiskit-aer
```

````python
from qiskit import transpile
from qiskit_aer import Aer

# Reuse the circuit from previous step (or create a new one)
backend = Aer.get_backend('qasm_simulator')
compiled = transpile(qc, backend)
job = backend.run(compiled, shots=1024)
result = job.result()
counts = result.get_counts()

print('Simulation counts:', counts)
````

## H. Optional: Visualize Results

Install plotting dependencies (if not present):

```bash
pip install matplotlib
```

````python
from qiskit.visualization import plot_histogram

plot_histogram(counts)
````

## I. Troubleshooting Checklist

If you face issues:

1. Confirm environment is active: `conda activate qiskit-env`
2. Upgrade pip: `python -m pip install --upgrade pip`
3. Reinstall Qiskit cleanly:
   ```bash
   pip uninstall -y qiskit qiskit-aer
   pip install qiskit qiskit-aer
   ```
4. Restart Jupyter kernel after installation changes.
5. Verify notebook uses the same Python environment shown in Step E.

## J. Success Criteria

Your installation is successful when:

- Qiskit imports without errors.
- Version is displayed correctly.
- Circuit simulation returns measurement counts (e.g., around 50/50 for `00` and `11` in Bell-state example).

You are now ready to build quantum circuits with Qiskit!
